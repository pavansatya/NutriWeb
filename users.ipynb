### Create derived features

The BMI values are categorised based on values given in the Centre of Disease Control website. We created two new features called `BMI_Category` which is a categorical variable that categorises the BMI values into **Underweight**, **Normal weight**, **Overweight**, and **Obese**. The second feature is `Caloric_Needs` which is a numerical variable that calculates the caloric needs based on the **Revised Harris-Benedict Equation**.

The Revised Harris-Benedict Equation is a widely used method to estimate daily calorie needs based on age, gender, weight, height, and activity level. This equation was developed in the 1910s by Dr. James Harris and Dr. Francis Benedict and has been revised over the years to account for changes in lifestyle and dietary habits.

The Revised Harris-Benedict Equation takes into consideration the basal metabolic rate (BMR), which is the number of calories the body needs to maintain basic functions such as breathing and circulation, and the physical activity level (PAL), which is the amount of energy expended during daily activities.

The formula for the Revised Harris-Benedict Equation is:

For men: BMR = 13.397 * weight(in kgs) + 4.799 * height(in cms) − 5.677 * age(in yrs) + 88.362

For women: BMR = 9.247 * weight(in kgs) + 3.098 * height(in cms) − 4.33 * age(in yrs) + 447.593

Once you have calculated the BMR using the appropriate formula, you can determine the daily calorie needs by multiplying the BMR by the appropriate physical activity level (PAL) factor:

- Sedentary : BMR x 1.2
- Moderate : BMR x 1.55
- Active : BMR x 1.9

```python
# BMI Category
def get_bmi_category(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif 18.5 <= bmi < 25:
        return 'Normalweight'
    elif 25 <= bmi < 30:
        return 'Overweight'
    else:
        return 'Obese'

data['BMI_Category'] = data['BMI'].apply(get_bmi_category)

# Caloric Needs (Harris-Benedict Equation)
def calculate_caloric_needs(row):
    if row['Gender'] == 'Male':
        bmr = 88.362 + (13.397 * row['Weight_kg']) + (4.799 * row['Height_cm']) - (5.677 * row['Age'])
    else:
        bmr = 447.593 + (9.247 * row['Weight_kg']) + (3.098 * row['Height_cm']) - (4.330 * row['Age'])
    
    if row['Physical_Activity_Level'] == 'Sedentary':
        return bmr * 1.2
    elif row['Physical_Activity_Level'] == 'Moderate':
        return bmr * 1.55
    else:
        return bmr * 1.9

data['Caloric_Needs'] = data.apply(calculate_caloric_needs, axis=1)
```

### Encode categorical variables

```python
from sklearn.preprocessing import LabelEncoder

df = pd.get_dummies(data, columns=['Gender', 'Disease_Type', 'Preferred_Cuisine', 'Allergies'])

label_encoder = LabelEncoder()
df['BMI_Category'] = label_encoder.fit_transform(df['BMI_Category'])
df['Severity'] = label_encoder.fit_transform(df['Severity'])
df['Physical_Activity_Level'] = label_encoder.fit_transform(df['Physical_Activity_Level'])
df['Dietary_Restrictions'] = label_encoder.fit_transform(df['Dietary_Restrictions'])
```
